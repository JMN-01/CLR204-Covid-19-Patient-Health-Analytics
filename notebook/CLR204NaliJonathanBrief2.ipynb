##### 3.1 Data Pre-processing

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# These are the references
# Pykes, K. (2025, January 15). Data Preprocessing: A Complete Guide with Python Examples. Datacamp. https://www.datacamp.com/blog/data-preprocessing
# Jain, D. (2026, February 7). Data Preprocessing in Data Mining. Geeksforgeeks.  https://www.geeksforgeeks.org/data-science/data-preprocessing-in-data-mining/

# Data preprocessing and cleaning for the COVID-19 patient dataset
# This loads the raw dataset
raw_df = pd.read_csv('data.csv')
df = raw_df.copy()

print("="*50)
print("Original Data Shape:", df.shape)

# This is for the Date Parsing & the delay inhospital visits after symptom onset calculation
df['sym_on'] = pd.to_datetime(df['sym_on'], errors='coerce')
df['hosp_vis'] = pd.to_datetime(df['hosp_vis'], errors='coerce')

df['diff_sym_hos'] = (df['hosp_vis'] - df['sym_on']).dt.days
df['diff_sym_hos'] = df['diff_sym_hos'].apply(lambda x: x if x >= 0 else np.nan)
df['diff_sym_hos'] = df['diff_sym_hos'].fillna(df['diff_sym_hos'].median())

# This is for Numerical & Binary Processing for the age and binary columns
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['age'] = df['age'].fillna(df['age'].median())

binary_cols = ['vis_wuhan', 'from_wuhan', 'death', 'recov']
for col in binary_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)

# This is for the label encoding of the categorical columns and text normalization
cat_cols = [
    'location', 'country', 'gender', 
    'symptom1', 'symptom2', 'symptom3', 
    'symptom4', 'symptom5', 'symptom6'
]

le = LabelEncoder()
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.lower()
        df[col] = df[col].replace({'nan': 'na', '': 'na', 'none': 'na', 'null': 'na'}).fillna('na')
        df[col] = le.fit_transform(df[col])

#  THis is for the make the final deduplicated feature selection (the 15 Core Features)
features = [
    'location', 'country', 'gender', 'age', 
    'vis_wuhan', 'from_wuhan', 'death', 'recov', 
    'symptom1', 'symptom2', 'symptom3', 'symptom4', 
    'symptom5', 'symptom6', 'diff_sym_hos'
]

cleaned_df = df[features].copy()

# Automatically save the preprocessed data file
cleaned_df.to_csv('cleaned_data.csv', index=False)
print("Preprocessed Data Shape:", cleaned_df.shape)
print(" ")
print("="*50)
print(" Missing Value Count After Preprocessing ")
print("="*50)
print(cleaned_df.isnull().sum())
print(" ")
print("="*50)
print("First 10 rows of Cleaned Dataset:")
print("="*50)
print(cleaned_df.head(10).to_string())

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import recall_score as rs, precision_score as ps, f1_score as fs
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix

# Same references were used throughout the evaluation metrics section
# These are the references for the boosted random forest and the evaluation metrics
# Aman, S. (2026, March 30). Evaluation Metrics in Machine Learning . Geeksforgeeks. https://www.geeksforgeeks.org/machine-learning/metrics-for-machine-learning-model/
# Peshkar, A. (2020). Covid-19-Patient-Health-Analytics. Github. https://github.com/Peshkar/Covid-19-Patient-Health-Analytics
# Scikit. (2026). 3.4. Metrics and scoring: quantifying the quality of predictions. https://scikit-learn.org/stable/modules/model_evaluation.html
# Scikit. (2026). DecisionTreeClassifier. https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html
# Scikit. (2026). train_test_split. https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html
# shahjamanbadhon.(2025). Machine-learning-practice. Github. https://github.com/shahjamanbadhon/machine-learning-practice/blob/main/Confusion_Matrix.ipynb
# W3Schools. (2026). Python Machine Learning - Confusion Matrix. https://www.w3schools.com/python/python_ml_confusion_matrix.asp
# Sharma, A. (2026, January 23). Understanding the Confusion Matrix in Machine Learning. Geeksforgeeks. https://www.geeksforgeeks.org/machine-learning/confusion-matrix-machine-learning/
# Galarnyk, M. (2025, August 4). Train Test Split: What It Means and How to Use It. Built In. https://builtin.com/data-science/train-test-split


tdata = pd.read_csv('train.csv')
# This is to calculate the diff_sym_hos from symptom onset and hospital visit dates
tdata['sym_on'] = pd.to_datetime(tdata['sym_on'], errors='coerce')
tdata['hosp_vis'] = pd.to_datetime(tdata['hosp_vis'], errors='coerce')
tdata['diff_sym_hos'] = (tdata['hosp_vis'] - tdata['sym_on']).dt.days
tdata['diff_sym_hos'] = tdata['diff_sym_hos'].apply(lambda x: x if x >= 0 else np.nan)
tdata['diff_sym_hos'] = tdata['diff_sym_hos'].fillna(tdata['diff_sym_hos'].median())

# This is for the numerical & Binary Processing
tdata['age'] = pd.to_numeric(tdata['age'], errors='coerce')
tdata['age'] = tdata['age'].fillna(tdata['age'].median())

binary_cols = ['vis_wuhan', 'from_wuhan', 'death', 'recov']
for col in binary_cols:
    if col in tdata.columns:
        tdata[col] = pd.to_numeric(tdata[col], errors='coerce').fillna(0).astype(int)

# This is for categorical text normalization & label encoding
cat_cols = ['location', 'country', 'gender', 'symptom1', 'symptom2', 'symptom3', 'symptom4', 'symptom5', 'symptom6']
le = LabelEncoder()
for col in cat_cols:
    if col in tdata.columns:
        tdata[col] = tdata[col].astype(str).str.strip().str.lower()
        tdata[col] = tdata[col].replace({'nan': 'na', '': 'na', 'none': 'na', 'null': 'na'}).fillna('na')
        tdata[col] = le.fit_transform(tdata[col])

# Model training adn evaluation#
X = tdata[['location', 'country', 'gender', 'age', 'vis_wuhan', 'from_wuhan', 'symptom1', 'symptom2', 'symptom3', 'symptom4', 'symptom5', 'symptom6', 'diff_sym_hos']]
Y = tdata['death']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=10)

# This initializes and fits the Support Vector Classifier
classifier = SVC(random_state=42)
classifier.fit(X_train, np.array(Y_train).reshape(Y_train.shape[0], 1).ravel())

# this is for the Evaluation of the model
pred = np.array(classifier.predict(X_test))

recall_svm = rs(Y_test, pred, zero_division=0)
precision_svm = ps(Y_test, pred, zero_division=0)
f1_svm = fs(Y_test, pred, zero_division=0)
ma_svm = classifier.score(X_test, Y_test)

print(" ")
print("="*70)
print(' Evaluation metrics for test dataset (Support Vector Machine) ')
print("="*70)
print('Recall Score: ', recall_svm)
print('Precision Score: ', precision_svm)
print('F1 Score: ', f1_svm)
print('Accuracy: ', ma_svm)

a = pd.DataFrame(Y_test)
a['pred'] = classifier.predict(X_test)
print('\n\tTable 3\n')
print(a.head())

# This Calculates and extracts the Confusion Matrix (TP, TN, FP, FN)
cm_svm = confusion_matrix(Y_test, pred)
TN_svm, FP_svm, FN_svm, TP_svm = cm_svm.ravel()

print(" ")
print("="*50)
print(' Confusion Matrix (Support Vector Machine) ')
print("="*50)
print(f"True Negatives (TN):          {TN_svm}")
print(f"False Positives (FP):         {FP_svm}")
print(f"False Negatives (FN):         {FN_svm}")
print(f"True Positives (TP):          {TP_svm}")

# This is for the plot
plt.bar(['Accuracy','F1 Score','Recall Score','Precision Score'],[ma_svm,f1_svm,recall_svm,precision_svm],color=['red','green','purple','orange'])
plt.plot([ma_svm,f1_svm,recall_svm,precision_svm],color='black')
plt.title('Evaluation Metrics for Support Vector Machine ',fontsize=12, pad=12)
plt.show()
```

##### 4.6 Decision Trees (Initial, 1, 10, 25 & 100 ) 

```python
import warnings
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)

# These are the references 
# Peshkar, A. (2020). Covid-19-Patient-Health-Analytics. Github. https://github.com/Peshkar/Covid-19-Patient-Health-Analytics
# Galarnyk, M. (2020). Decision Trees Visualization. Github. https://github.com/mGalarnyk/Python_Tutorials/blob/master/Sklearn/CART/Visualization/DecisionTreesVisualization.ipynb
# Scikit. (n.d.). Understanding the decision tree structure. https://scikit-learn.org/stable/auto_examples/tree/plot_unveil_tree_structure.html
# Scikit. (n.d.). Preprocessing data. https://scikit-learn.org/stable/modules/preprocessing.html

# The is to load and preprocess the data, including calculating the difference between symptom onset and hospital visit dates, handling missing values, and encoding categorical variables.
dataset = pd.read_csv("train.csv")

# This is the code for cleaning, formating and categorizing of the data
dataset["sym_on"] = pd.to_datetime(dataset["sym_on"], errors="coerce")
dataset["hosp_vis"] = pd.to_datetime(dataset["hosp_vis"], errors="coerce")
dataset["diff_sym_hos"] = (dataset["hosp_vis"] - dataset["sym_on"]).dt.days
dataset["diff_sym_hos"] = dataset["diff_sym_hos"].apply(
    lambda x: x if x >= 0 else np.nan
)
dataset["diff_sym_hos"] = dataset["diff_sym_hos"].fillna(
    dataset["diff_sym_hos"].median()
)
dataset["age"] = pd.to_numeric(dataset["age"], errors="coerce")
dataset["age"] = dataset["age"].fillna(dataset["age"].median())

binary_cols = ["vis_wuhan", "from_wuhan", "death", "recov"]
for col in binary_cols:
  if col in dataset.columns:
    dataset[col] = (
        pd.to_numeric(dataset[col], errors="coerce").fillna(0).astype(int)
    )

cat_cols = [
    "location",
    "country",
    "gender",
    "symptom1",
    "symptom2",
    "symptom3",
    "symptom4",
    "symptom5",
    "symptom6",
]
le = LabelEncoder()
for col in cat_cols:
  if col in dataset.columns:
    dataset[col] = dataset[col].astype(str).str.strip().str.lower()
    dataset[col] = (
        dataset[col]
        .replace({"nan": "na", "": "na", "none": "na", "null": "na"})
        .fillna("na")
    )
    dataset[col] = le.fit_transform(dataset[col])

X = dataset[
    [
        "location",
        "country",
        "gender",
        "age",
        "vis_wuhan",
        "from_wuhan",
        "symptom1",
        "symptom2",
        "symptom3",
        "symptom4",
        "symptom5",
        "symptom6",
        "diff_sym_hos",
    ]
]
Y = dataset["death"]

# This is where the coloumn names are changed (Country = x1)
feature_labels = [f"X{i}" for i in range(X.shape[1])]

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=10, stratify=Y
)

# This is the code for Hyperparameter Tuning using GridSearchCV & Pipeline 
selector = SelectKBest(score_func=f_classif)
clf = RandomForestClassifier(random_state=42)

pipeline = Pipeline(
    [("feature_selection", selector), ("classification", clf)]
)

param_grid = {
    "feature_selection__k": [5, 8, 10, "all"],
    "classification__n_estimators": [50, 100],
    "classification__criterion": ["gini", "entropy"],
    "classification__max_depth": [3, 5, 10, None],
    "classification__min_samples_split": [2, 5, 10],
}

cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=cv_strategy,
    scoring="f1",
    n_jobs=-1,
)

grid_search.fit(X_train, Y_train)

# This is the iinitial decision tree
# This is for the decision tree visualization (Initial Decision Tree with max_depth=2)
# Agarwal,Y. (2025, July 23). How to Visualize a Decision Tree from a Random Forest. https://www.geeksforgeeks.org/machine-learning/ways-to-visualize-individual-decision-trees-in-a-random-forest/
dt_clean = DecisionTreeClassifier(max_depth=2, random_state=42)
dt_clean.fit(X, Y)

plt.figure(figsize=(10, 6), dpi=150)
plot_tree(
    dt_clean,
    feature_names=feature_labels,
    filled=True,
    rounded=True,
    fontsize=10,
)
plt.title("Decision Tree (Initial) ", fontsize=13, pad=12)
plt.tight_layout()
plt.show()

# Navlani, A. (2026, January 16). Decision Tree Classification in Python Tutorial. https://www.datacamp.com/tutorial/decision-tree-classification-python
# Train one Random Forest with 101 trees, then plot different individual trees from inside it
rf = RandomForestClassifier(n_estimators=101, max_depth=2, random_state=10)
rf.fit(X_train, Y_train)

tree_indices = [1, 10, 25, 100]

for idx in tree_indices:
  plt.figure(figsize=(10, 6), dpi=150)
  plot_tree(
      rf.estimators_[idx],
      feature_names=feature_labels,
      filled=True,
      rounded=True,
      fontsize=10,
  )
  plt.title(
      f"Decision Tree (n_estimator {idx})",
      fontsize=13,
      pad=12,
  )
  plt.tight_layout()
  plt.show()
```

## **6. Hyperparameter optimisation:**

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd
import numpy as np
# Singh, T. (2026, June 26). Hyperparameter Tuning. Geeksforgeeks. https://www.geeksforgeeks.org/machine-learning/hyperparameter-tuning/
# lakshaymbnwg. (2025, July 23). Performing Feature Selection with gridsearchcv in Sklearn. https://www.geeksforgeeks.org/machine-learning/performing-feature-selection-with-gridsearchcv-in-sklearn/
# Scikit. (n.d.). GridSearchCV. https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html
# Scikit. (n.d.). Random Forest Classifier. https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html
#https://scikit-learn.org/stable/modules/model_evaluation.html
# Scikit. (n.d.). 3.2. Tuning the hyper-parameters of an estimator. https://scikit-learn.org/stable/modules/grid_search.html
# Scikit. (n.d.). 3.4. Metrics and scoring: quantifying the quality of predictions. https://scikit-learn.org/stable/modules/model_evaluation.html


# This is for defining the features and target variable, splitting the dataset into training and testing sets, and setting up the Random Forest Classifier with hyperparameter tuning using GridSearchCV.
X = tdata[['location', 'country', 'gender', 'age', 'vis_wuhan', 'from_wuhan', 'symptom1', 'symptom2', 'symptom3', 'symptom4', 'symptom5', 'symptom6', 'diff_sym_hos']]
Y = tdata['death']
X_train, X_test, Y_train, Y_test = train_test_split( X, Y, test_size=0.2, random_state=42, stratify=Y )
# This is for the base classifier
rf = RandomForestClassifier(random_state=42)

# this is for the parameters 
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [2, 5, 10, None],
    'min_samples_leaf': [1, 2, 4],
    'min_samples_split': [2, 5],
    'criterion': ['gini', 'entropy']
}

# Thsi sets up the GridSearchCV 
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    scoring='accuracy',
    n_jobs=-1
)

# Fit the grid search to your training data
grid_search.fit(X_train, Y_train)

# This extract and display the Optimal hyperparameters returned by grid search
hyperparameter_data = {
    "Parameters": list(grid_search.best_params_.keys()),
    "Value": list(grid_search.best_params_.values())
}
hyperparameter_df = pd.DataFrame(hyperparameter_data)
print("="*50)
print("Optimal hyperparameters returned by grid search.")
print("="*50)
print(hyperparameter_df.to_string(index=False))

```
